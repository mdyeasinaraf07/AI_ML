```python

import pandas as pd
import numpy as np

# 1. Create a DataFrame
df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Rakib", "Nila", "Rafi"],
    "Age": [20, 21, 19, 22, 20],
    "Study_Hours": [3, 5, 7, np.nan, 2],
    "Attendance": [80, 90, 95, 70, 60],
    "Marks": [65, 75, 95, 55, 45]
})

print("Original DataFrame:\n", df)


# 2. View Data
print("First 5 Rows:\n", df.head())
print("First 3 Rows:\n", df.head(3))
print("Last 5 Rows:\n", df.tail())
print("Random Rows:\n", df.sample(2))


# 3. DataFrame Information
print("Shape:", df.shape)
print("Columns:", df.columns)
print("Index:", df.index)
print("Data Types:\n", df.dtypes)
df.info()
print("Summary:\n", df.describe())


# 4. Select Columns
print("One Column:\n", df["Marks"])
print("Multiple Columns:\n", df[["Name", "Marks"]])


# 5. Select Rows and Cells
print("First Row:\n", df.iloc[0])
print("First Two Rows:\n", df.iloc[:2])
print("Cell Value:", df.loc[0, "Marks"])
print("First Row, Marks Column:", df.iloc[0, 4])


# 6. Filter Data
print("Marks > 70:\n", df[df["Marks"] > 70])
print("Age >= 20:\n", df[df["Age"] >= 20])
print("Multiple Conditions:\n",
      df[(df["Marks"] >= 60) & (df["Attendance"] >= 80)])
print("OR Condition:\n",
      df[(df["Marks"] > 90) | (df["Attendance"] > 90)])
print("Using isin:\n", df[df["Age"].isin([19, 20])])


# 7. Minimum, Maximum and Statistics
print("Minimum:", df["Marks"].min())
print("Maximum:", df["Marks"].max())
print("Sum:", df["Marks"].sum())
print("Mean:", df["Marks"].mean())
print("Median:", df["Marks"].median())
print("Mode:\n", df["Marks"].mode())
print("Standard Deviation:", df["Marks"].std())
print("Variance:", df["Marks"].var())
print("Count of Non-Missing Values:", df["Marks"].count())
print("Unique Value Count:", df["Marks"].nunique())
print("Unique Values:", df["Marks"].unique())
print("Value Counts:\n", df["Marks"].value_counts())


# 8. Sorting
print("Ascending:\n", df.sort_values("Marks"))
print("Descending:\n", df.sort_values("Marks", ascending=False))
print("Sort by Multiple Columns:\n",
      df.sort_values(["Age", "Marks"]))


# 9. Missing Values
print("Missing Values:\n", df.isna().sum())
print("Missing Value Mask:\n", df.isnull())
print("Rows with Missing Values:\n", df[df.isna().any(axis=1)])

# Fill missing Study_Hours with its mean
df["Study_Hours"] = df["Study_Hours"].fillna(
    df["Study_Hours"].mean()
)

print("After Fill:\n", df)


# 10. Remove Missing Values
df_copy = df.copy()
print("Drop Rows with Missing Values:\n", df_copy.dropna())
print("Drop Columns with Missing Values:\n", df_copy.dropna(axis=1))


# 11. Duplicate Values
print("Duplicate Rows:\n", df.duplicated())
print("Duplicate Count:", df.duplicated().sum())
print("After Removing Duplicates:\n", df.drop_duplicates())


# 12. Create and Modify Columns
df["Marks_Doubled"] = df["Marks"] * 2
df["Result"] = np.where(df["Marks"] >= 60, "Pass", "Fail")
df["Study_Hours_Plus_One"] = df["Study_Hours"] + 1
print(df)


# 13. Rename and Drop Columns
renamed_df = df.rename(columns={"Marks": "Score"})
print("Renamed:\n", renamed_df)

print("Drop Column:\n", df.drop(columns=["Marks_Doubled"]))
print("Drop First Row:\n", df.drop(index=0))


# 14. Apply a Function
df["Marks_Plus_5"] = df["Marks"].apply(lambda x: x + 5)
print(df[["Name", "Marks", "Marks_Plus_5"]])


# 15. Replace Values
print(df["Result"].replace({"Pass": 1, "Fail": 0}))
print(df["Name"].replace({"Rahim": "Rahim Ahmed"}))


# 16. Change Data Types
print(df["Age"].astype(float))
print(pd.to_numeric(
    pd.Series(["80", "90", "unknown"]),
    errors="coerce"
))


# 17. String Methods
names = df["Name"]
print(names.str.lower())
print(names.str.upper())
print(names.str.strip())
print(names.str.contains("a", case=False))


# 18. GroupBy and Aggregation
group_df = pd.DataFrame({
    "Department": ["CSE", "CSE", "EEE", "EEE", "CSE"],
    "Marks": [80, 90, 70, 85, 75]
})

print(group_df.groupby("Department")["Marks"].mean())
print(group_df.groupby("Department")["Marks"].agg(
    ["mean", "min", "max", "count"]
))


# 19. Value Counts and Percentages
print(df["Result"].value_counts())
print(df["Result"].value_counts(normalize=True))


# 20. Correlation
print("Correlation:\n",
      df[["Age", "Study_Hours", "Attendance", "Marks"]].corr())


# 21. Combine DataFrames
df1 = pd.DataFrame({"ID": [1, 2], "Name": ["Rahim", "Karim"]})
df2 = pd.DataFrame({"ID": [3, 4], "Name": ["Nila", "Rafi"]})

print("Concatenate Rows:\n", pd.concat([df1, df2], ignore_index=True))


# 22. Merge / Join Data
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Rahim", "Karim", "Nila"]
})

scores = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 75, 90]
})

print("Merge:\n", pd.merge(students, scores, on="ID"))
print("Left Merge:\n", pd.merge(students, scores, on="ID", how="left"))


# 23. Pivot Table
print(pd.pivot_table(
    group_df,
    values="Marks",
    index="Department",
    aggfunc="mean"
))


# 24. Reset and Set Index
indexed_df = df.set_index("Name")
print("Set Index:\n", indexed_df)
print("Reset Index:\n", indexed_df.reset_index())


# 25. Transpose
print("Transpose:\n", df.T)


# 26. Read and Write CSV
# Uncomment when students.csv exists
# data = pd.read_csv("students.csv")
# data.to_csv("cleaned_students.csv", index=False)


# 27. Read and Write Excel
# Requires a compatible Excel engine, such as openpyxl
# data = pd.read_excel("students.xlsx")
# data.to_excel("cleaned_students.xlsx", index=False)


# 28. Read JSON
# data = pd.read_json("students.json")
# data.to_json("output.json", orient="records", indent=2)


# 29. Convert to NumPy Array
marks_array = df["Marks"].to_numpy()
print("NumPy Array:", marks_array)

# DataFrame to dictionary
print(df.to_dict(orient="records"))


# 30. Date and Time Methods
dates = pd.Series(pd.to_datetime([
    "2026-01-10", "2026-02-15", "2026-03-20"
]))

print("Year:", dates.dt.year)
print("Month:", dates.dt.month)
print("Day:", dates.dt.day)
print("Day of Week:", dates.dt.dayofweek)
```
