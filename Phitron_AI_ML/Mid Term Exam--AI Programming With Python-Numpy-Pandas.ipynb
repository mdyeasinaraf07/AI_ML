```python
# You must run this code and use this dataset for all questions

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)

```

# Q1:

```python
print(f"Employees Ages:\n{emp_ages}\n")
print(f"Employees Incomes:\n{emp_income}\n")
print(f"For Employees Ages:\nData type: {emp_ages.dtype}\nNumber of dimensions: {emp_ages.ndim}\nShape: {emp_ages.shape}\nTotal number of elements: {emp_ages.size}\n")
print(f"For Employees Incomes:\nData type: {emp_income.dtype}\nNumber of dimensions: {emp_income.ndim}\nShape: {emp_income.shape}\nTotal number of elements: {emp_income.size}\n")

```

# Q2:


```python
emp_maximum_salary = np.max(emp_income)
emp_minimum_salary = np.min(emp_income)

emp_average_salary = np.mean(emp_income)
emp_average_age = np.mean(emp_ages)

print(f"Employees Maximum Salary: {emp_maximum_salary}\nEmployees Minimum Salary: {emp_minimum_salary}\n")
print(f"Employees Average Salary: {emp_average_salary}\n")
print(f"Employees Average Age: {emp_average_age}\n")

```

# Q3:

```python
filter_values = (emp_ages >= 25) & (emp_ages <= 40) & (emp_income > 50000)
count = np.sum(filter_values)

print(f"Filtered values:\n{emp_ages[filter_values]}\n")
print(f"Number of Employees Satisfy this Condition: {count}\n")

```

# Q4:

```python
average_age = np.mean(emp_ages)
employees_adjusted_age = np.where(emp_ages <  average_age, emp_ages + 4, emp_ages + 2)

print(f"Original ages:\n{emp_ages}\n")
print(f"Updated ages:\n{employees_adjusted_age}\n")
print(f"Average age: {average_age}\n")


```

# Q5:

```python
tech_roles_indx = np.where((emp_roles == "Software Developer") | (emp_roles == "Backend Developer") | (emp_roles == "Digital Marketer")) [0]
tech_roles_salary = emp_income[tech_roles_indx]
total_tech_roles_salary = np.sum(tech_roles_salary)

tech_roles_salaries_increase = (tech_roles_salary * 1.08)
total_tech_roles_salaries_increase = np.sum(tech_roles_salaries_increase)

old_total_salary_expense = np.sum(tech_roles_salary)
new_total_salary_expense = np.sum(tech_roles_salaries_increase)

overall_increase_expense_after_the_raise = (new_total_salary_expense - old_total_salary_expense)

print(f"Old Total Salary Expense: {old_total_salary_expense}\n")
print(f"New Total Salary Expense: {new_total_salary_expense}\n")
print(f"Difference: {overall_increase_expense_after_the_raise}\n")

```
