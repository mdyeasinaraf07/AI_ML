Answer to the Question number: 2

```python
x, y = map(int, input().split())

def compute_ratio(x, y):
  try:
    temp  = (x + y) / (x * y)
  except Exception:
    return "Error: Division not possible"
  else:
    return temp

ans = compute_ratio(x, y)
print(ans)
```

```python
with open("./sample_data/data.txt", "r") as file:
  txt = file.read().split()
  longest_word = txt[0]
  for i in txt:
    if(len(i) > len(longest_word)):
      longest_word = i

  print(longest_word)
```

Answer to the Question number: 5

```python
txt = input().lower()
reverse_txt = txt[::-1]
if(txt == reverse_txt):
  print(f"Palindrome")
else:
  print(f"Not Palindrome")
```
