```python

import numpy as np

# 1. Create a NumPy Array
arr = np.array([10, 20, 30, 40, 50])
print("Array:", arr)

# 2. Array Properties
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Dimensions:", arr.ndim)
print("Data Type:", arr.dtype)

# 3. Minimum and Maximum
print("Minimum:", arr.min())
print("Maximum:", arr.max())
print("Minimum (np.min):", np.min(arr))
print("Maximum (np.max):", np.max(arr))

# 4. Sum and Average
print("Sum:", arr.sum())
print("Mean / Average:", arr.mean())
print("Standard Deviation:", arr.std())
print("Variance:", arr.var())

# 5. Index of Minimum and Maximum
print("Index of Minimum:", arr.argmin())
print("Index of Maximum:", arr.argmax())

# 6. Sorting
unsorted = np.array([40, 10, 50, 20, 30])
print("Sorted:", np.sort(unsorted))

# 7. Create Special Arrays
print("Zeros:", np.zeros(5))
print("Ones:", np.ones(5))
print("Full:", np.full(4, 7))
print("Infinity:", np.full(4, np.inf))
print("Identity Matrix:\n", np.eye(3))

# 8. Evenly Spaced Values
print("Arange:", np.arange(0, 10, 2))
print("Linspace:", np.linspace(0, 10, 5))

# 9. Indexing
print("First Element:", arr[0])
print("Last Element:", arr[-1])
print("Third Element:", arr[2])

# 10. Slicing
print("First Three:", arr[:3])
print("From Index 2:", arr[2:])
print("Index 1 to 3:", arr[1:4])
print("Every Second Element:", arr[::2])
print("Reverse Array:", arr[::-1])

# 11. Change Array Values
arr2 = np.array([10, 20, 30, 40, 50])
arr2[0] = 100
arr2[1:4] = 0
print("Updated Array:", arr2)

# 12. Boolean Filtering
print("Values > 25:", arr[arr > 25])
print("Even Values:", arr[arr % 2 == 0])
print("Values < 30:", arr[arr < 30])

# 13. Arithmetic Operations
print("Add:", arr + 5)
print("Subtract:", arr - 5)
print("Multiply:", arr * 2)
print("Divide:", arr / 2)
print("Square:", arr ** 2)

# 14. Useful Mathematical Functions
print("Absolute:", np.abs([-10, -20, 30]))
print("Square Root:", np.sqrt(arr))
print("Rounded:", np.round([1.234, 2.678], 2))
print("Exponential:", np.exp([1, 2]))
print("Log:", np.log([1, np.e]))

# 15. Two-Dimensional Array
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print("Matrix:\n", matrix)
print("Shape:", matrix.shape)
print("First Row:", matrix[0])
print("Second Row:", matrix[1])
print("First Column:", matrix[:, 0])
print("Second Column:", matrix[:, 1])
print("Element at Row 1, Column 2:", matrix[1, 2])

# 16. Reshape
numbers = np.arange(1, 7)
print("Reshaped:\n", numbers.reshape(2, 3))
print("Reshape with -1:\n", numbers.reshape(3, -1))

# 17. Axis Operations
print("Column Sum:", matrix.sum(axis=0))
print("Row Sum:", matrix.sum(axis=1))
print("Column Mean:", matrix.mean(axis=0))
print("Row Maximum:", matrix.max(axis=1))

# 18. Transpose
print("Transpose:\n", matrix.T)

# 19. Join and Split Arrays
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("Concatenate:", np.concatenate((a, b)))
print("Split:", np.split(np.array([1, 2, 3, 4]), 2))

# 20. Unique Values and Counts
values = np.array([1, 2, 2, 3, 3, 3])
print("Unique:", np.unique(values, return_counts=True))

# 21. Random Arrays
print("Random 1D:", np.random.rand(3))
print("Random 2D:", np.random.rand(2, 3))
print("Random Integers:", np.random.randint(1, 10, size=5))

# 22. Find NaN and Infinity
special = np.array([1.0, np.nan, np.inf, 4.0])
print("Is NaN:", np.isnan(special))
print("Is Finite:", np.isfinite(special))
print("Replace NaN:", np.nan_to_num(special))

# 23. Clip Values
print("Clip:", np.clip(arr, 20, 40))

# 24. Check Conditions
print("Any > 40:", np.any(arr > 40))
print("All > 0:", np.all(arr > 0))
print("Count > 25:", np.count_nonzero(arr > 25))

# 25. Celsius to Fahrenheit
celsius = np.array([10, 20, 30, 40, 50])
fahrenheit = (celsius * 1.8) + 32
print("Fahrenheit:", fahrenheit)
```
