```python
import numpy as np

temperatures_arr = [10, 20, 30, 40, 50]
celsius_arr = np.array(temperatures_arr)

fahrenheit_arr = (celsius_arr * 1.8) + 32
fahrenheit_arr = fahrenheit_arr.astype(np.int32)

print(fahrenheit_arr)
```

```python
lap_times_arr = [20.5, 25.4, 30.1, 15.6, 50.10]
lap_arr = np.array(lap_times_arr)

print(f"Shape: {lap_arr.shape}\nSize: {lap_arr.size}\nData type: {lap_arr.dtype}")

lap_times_minutes = lap_arr / 60

print(lap_times_minutes)
```

## Create a 4×4 matrix from 0–15. Then , extract:
the first row


the last column


a 2×2 subarray from the center


```python
matrix = np.arange(0, 16).reshape(4, 4)
print(matrix)

# first_row_matrix = matrix[0]
first_row_matrix = matrix.take(0, axis = 0)

# last_col_matrix = matrix[:, 3]
last_col_matrix = matrix.take(3, axis = 1)

center_sub_arr = matrix[1:3, 1:3]

print(f"The first row of matrix: {first_row_matrix}")
print(f"The last column of matrix: {last_col_matrix}")
print(f"The center sub array of matrix: {center_sub_arr}")
```
